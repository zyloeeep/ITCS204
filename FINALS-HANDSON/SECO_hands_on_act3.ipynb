```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul"]
for i in range(streams.shape[0]):
    plt.plot(months, streams[i], label=music[i], marker='o')

plt.title("Monthly Streams per Song")
plt.xlabel("Month")
plt.ylabel("Streams")
plt.gcf().set_size_inches(10, 7)
plt.legend()
plt.show()
```

```python
rev = np.arange(len(music))

plt.figure(figsize=(10,5))
plt.bar(rev, revenue[:, -1], color='skyblue')
plt.xticks(rev, music, rotation=45)
plt.ylabel("Revenue (1000s)")
plt.title("Revenue of Songs in Last Month")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.imshow(ratings, cmap="YlGnBu", aspect='auto') 
plt.colorbar(label="Rating") 

plt.xticks(np.arange(7), ["W1","W2","W3","W4","W5","W6","W7"])
plt.yticks(np.arange(len(music)), music)
plt.title("Weekly Ratings of Songs")
plt.xlabel("Week")
plt.ylabel("Song")
plt.show()
```

```python
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul"]
plt.figure(figsize=(10,7))
for i in range(comments.shape[0]):
    plt.scatter(months, comments[i], label=music[i], s=100)

plt.title("Monthly Comments per Song")
plt.xlabel("Month")
plt.ylabel("Number of Comments")
plt.legend()
plt.show()
```
