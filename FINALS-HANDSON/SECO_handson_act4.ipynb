```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
world_happiness = pd.read_csv("C:/Users/Cielo Seco/Downloads/archive/world_happiness_report.csv")
```

```python
columns_to_use = ['Happiness Score', 
                  'Economy (GDP per Capita)', 
                  'Health (Life Expectancy)',
                  'Freedom', 
                  'Trust (Government Corruption)']

data_subset = world_happiness[columns_to_use].dropna()
```

```python
# 1. Economy vs Happiness Score

plt.figure(figsize=(10,5))
sns.scatterplot(data=data_subset, x='Economy (GDP per Capita)', y='Happiness Score')
plt.title("Economy (GDP per Capita) vs Happiness Score")
plt.xlabel("Economy (GDP per Capita)")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()
```

As the GDP per capital increases, the happiness score tends to rise as well. This suggests that citizens in countries with stronger economies generally report higher levels of happiness.

```python
# 2. Health (Life Expectancy) vs Happiness Score

plt.figure(figsize=(10,5))
sns.scatterplot(data=data_subset, x='Health (Life Expectancy)', y='Happiness Score')
plt.title("Health (Life Expectancy) vs Happiness Score")
plt.xlabel("Health (Life Expectancy)")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()
```

As life expectancy increases, the happiness score also tends to go up. This indicates that populations with better health outcomes and longer life expectancies generally perceive themselves as happier.

```python
# 3. Freedom vs Happiness Score

plt.figure(figsize=(10,5))
sns.scatterplot(data=data_subset, x='Freedom', y='Happiness Score')
plt.title("Freedom vs Happiness Score")
plt.xlabel("Freedom")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()
```

As the freedom to make life choices increases, the happiness score generally trends upward. This implies that citizens who feel they have the autonomy to direct their own lives tend to report higher levels of happiness.

```python
# 4. Trust (Government Corruption) vs Happiness Score

plt.figure(figsize=(10,5))
sns.scatterplot(data=data_subset, x='Trust (Government Corruption)', y='Happiness Score')
plt.title("Trust (Government Corruption) vs Happiness Score")
plt.xlabel("Trust (Government Corruption)")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()
```

It shows that countries with higher public trust in their government and lower perceived corruption report higher levels of happiness.

The data indicates that a country's happiness is closely linked to its economy, health, and level of freedom. The clearest trends show that people in wealthier nations with longer life expectancies usually report the highest levels of happiness. Additionally, having the freedom to make life choices and living in a society with less government corruption are also key factors that boost happiness. Overall, while financial stability and physical health are the biggest drivers, personal liberty and trust in government are also essential for a happy population.
